## ¿Cómo abrimos y reproducimos un video en OpenCV?

```python
ruta_video = r"C:\Users\lfiza\OneDrive\Documents\Personal\UTEC\2026-2\VCRP\VCRP - Luis\Teoria\Semana 3\Actividad Clase\video_caminando.MOV" 
video = cv.VideoCapture(ruta_video)

while video.isOpened():
    ret, frame = video.read()

    '''
    - ret es un booleano -> si es True entonces si se leyó el video correctamente, sino False
    - frame parece que es una lista de frames porque en la documentción dice "capture frame-by-frame"
    '''

    if not ret:
        print("No se ha podido leer bien el video")
        break

    '''
    imshow() tiene dos parametros: 
    - un const string& -> ¿que nombre le ponemos a la ventana en la que se va a mostrar el video?
    - un const ogl::texture2D -> la imagen o frame pero tienes que indicarle: ¿a qué modelo de color lo convertimos?
    '''
    
    '''
    - cv.COLOR_BGR2Lab -> lo convertimos a CIE l*a*b* (modelo de color más cercano al ojo humano) -> ¿Por qué sale sólo amarillo y azul?
    - cv.COLOR_RGB2LAB -> lo convertimos a CIE l*a*b* (modelo de color más cercano al ojo humano) -> ¿Por qué ahora sólo amarillo, verde y azul?   
    '''

    cie_lab_BGR = cv.cvtColor(frame, cv.COLOR_BGR2LAB)
    cie_lab_RGB = cv.cvtColor(frame, cv.COLOR_RGB2LAB)

    cv.imshow("Video Caminando", cie_lab_RGB)
    if cv.waitKey(1) == ord('q'):
        break

video.release()
cv.destroyAllWindows()
```
